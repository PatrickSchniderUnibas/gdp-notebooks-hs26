## Aufgabe 3 (4 Punkte)
Schreiben Sie eine Methode `numberOfPrimes` welche zwei ganzzahlige Parameter `lo` und `hi` entgegennimmt und zurückgibt, wieviele Primzahlen zwischen `lo` und `hi` sind. Zur Erinnerung: eine Zahl ist eine Primzahl, wenn sie nur durch 1 und sich selbst teilbar ist. Für uns ist 1 keine Primzahl. Beim Aufruf `numberOfPrimes(2, 11)` müsste die Methode also `5` zurückgeben.

````java
class Numbers {

    // Ihre Lösung

    public static void main(String[] args) {
        // Aufruf für lo = 2 und hi = 11
        System.out.println(numberOfPrimes(2, 11));
    }

}
````

````java
Numbers.main(new String[0]);
````

#### Lösung

````java
class Numbers {

    public static boolean isPrime(int n) {
        if (n == 1) {return false;}
        for (int i = 2; i <= n/2; i++) {
            if (n % i == 0) {return false;}
        }
        return true;
    }

    public static int numberOfPrimes(int lo, int hi) {
        int num = 0;
        for (int i = lo; i <= hi; i++) {
            if (isPrime(i)) {
                num = num + 1;
            }
        }
        return num;
    }

    public static void main(String[] args) {
        // Aufruf für lo = 2 und hi = 11
        System.out.println(numberOfPrimes(2, 11));
    }

}
````

````java
Wedge.main(new String[0]);
````

#### Lösung

````java
class Wedge {

    public static void drawWedge(int n) {
        for (int i = n; i >= 0; i--) {
            for (int j = 0; j < i; j++) {
                System.out.print("*");
            }
            System.out.println("");
        }
    }

    public static void main(String[] args) {
        drawWedge(5);
    }

}
````

* wenn $n$ ungerade ist, werd der Prozess mit $3n+1$ fortgeführt;
* der Prozess ist beendet, wenn die Zahl $1$ erreicht wird.

Ihre Methode soll alle Zahlen im Prozess ausgeben. Beim Aufruf `collatz(6)` sollen also die Zahlen `6, 3, 10, 5, 16, 8, 4, 2, 1` ausgegeben werden.

#### Lösung

````java
class Recursion {

    public static void collatz(int n) {
        System.out.println(n);
        if (n <= 1) {
            return;
        }
        if (n % 2 == 0) {
            collatz(n / 2);
        }
        else {
            collatz(3*n + 1);
        }
    }

    public static void main(String[] args) {
        collatz(6);
    }
}
````

````java
class Crossings{

    // Diese Methode sollen Sie nicht verändern
    public static boolean cross(double x1, double y1, double x2, double y2, double x3, double y3, double x4, double y4) {
        if (Math.signum(((y2 - y1) * (x3 - x2)) - ((x2 - x1) * (y3 - y2))) != Math.signum(((y2 - y1) * (x4 - x2)) - ((x2 - x1) * (y4 - y2))) &&
            Math.signum(((y4 - y3) * (x2 - x4)) - ((x4 - x3) * (y2 - y4))) != Math.signum(((y4 - y3) * (x1 - x4)) - ((x4 - x3) * (y1 - y4)))) {
            return true;
        }
        return false;
    }


    public static boolean crossingFamily(double[][][] segments) {
        for (int i = 0; i < segments.length; i++) {
            for (int j = i+1; j < segments.length; j++) {
                if (!cross(segments[i][0][0], segments[i][0][1], segments[i][1][0], segments[i][1][1], segments[j][0][0], segments[j][0][1], segments[j][1][0], segments[j][1][1])) {
                    return false;
                }
            }
        }
        return true;
    }

    public static void main(String[] args) {
        double[] p1 = new double[]{-1.72, -0.18};
        double[] p2 = new double[]{1.84, 3.68};
        double[] p3 = new double[]{-2.38, 3.98};
        double[] p4 = new double[]{1.66, -0.48};
        double[] p5 = new double[]{0.08, 5.38};
        double[] p6 = new double[]{-0.32, 0.36};
        double[] p7 = new double[]{3.44, 2.74};
        double[] p8 = new double[]{-7-38, 1.98};

        double[][] s1 = new double[][]{p1, p2};
        double[][] s2 = new double[][]{p3, p4};
        double[][] s3 = new double[][]{p5, p6};
        double[][] s4 = new double[][]{p7, p8};

        double[][][] segments = {s1, s2, s3, s4};
        System.out.println(crossingFamily(segments));
    }
}
````
